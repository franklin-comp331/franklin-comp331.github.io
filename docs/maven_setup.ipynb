# Maven Setup Guide

Maven builds Java projects, downloads libraries, and runs tests. You will use it throughout Labs 1-5 for the `comp331-java-agent` repository.

For the authoritative installation instructions and current downloads, use the [official Apache Maven installation page](https://maven.apache.org/install.html).

## 1. Confirm Java Is Installed

Maven requires a JDK. Install JDK 17 or later using the [Tool Setup Guide](tool_setup.ipynb), then confirm it is available:

```bash
java -version
```

If this command fails, install or configure Java before installing Maven.

## 2. Install Maven

Choose the instructions for your operating system. If your institution or development environment provides Maven already, skip to the verification step.

### macOS

With [Homebrew](https://brew.sh/):

```bash
brew install maven
```

### Windows

With [winget](https://learn.microsoft.com/windows/package-manager/winget/):

```powershell
winget install Apache.Maven
```

Close and reopen PowerShell after installation. If `winget` is unavailable, follow the manual Windows instructions on the [official installation page](https://maven.apache.org/install.html).

### Linux

On Debian or Ubuntu:

```bash
sudo apt update
sudo apt install maven
```

For other distributions or to install a specific version, use the [official installation page](https://maven.apache.org/install.html).

## 3. Verify the Installation

Open a new terminal and run:

```bash
mvn -version
```

The output should report an Apache Maven version and the Java version Maven will use. If the command is not found, reopen the terminal. Manual installations may also require adding Maven's `bin` directory to your `PATH`; use the official instructions for your operating system.

## 4. Confirm Maven Can Reach Maven Central

Maven downloads project dependencies from remote repositories, usually [Maven Central](https://central.sonatype.com/). From a Maven project directory containing `pom.xml`, run:

```bash
mvn clean test
```

On the first run, Maven may print `Downloading from central` messages while it stores dependencies in `~/.m2/repository` (macOS/Linux) or `%USERPROFILE%\.m2\repository` (Windows). A final `BUILD SUCCESS` confirms that Maven, Java, and repository access are working. Later runs are usually faster because dependencies are cached locally.

If a course project does not yet have tests, use this command to test dependency resolution directly:

```bash
mvn dependency:resolve
```

Connection errors such as `Unknown host`, `Connection timed out`, or `PKIX path building failed` usually indicate a network, proxy, firewall, or institutional certificate issue. Check whether you can open [Maven Central](https://central.sonatype.com/) in a browser, then ask campus IT for the Maven proxy or certificate settings if the problem persists. Do not disable TLS certificate verification.

## 5. Create the Lab Project

From the parent directory where you keep course repositories, create a minimal project:

```bash
mvn archetype:generate -DgroupId=edu.comp331 -DartifactId=comp331-java-agent -DarchetypeArtifactId=maven-archetype-quickstart -DinteractiveMode=false
cd comp331-java-agent
```

This creates `pom.xml`, `src/main/java`, and `src/test/java`. You may instead create the Maven project through IntelliJ IDEA or VS Code if it produces the same standard structure.

## 6. Common Commands

Run these commands from the directory containing `pom.xml`:

```bash
mvn compile       # compile the application
mvn test          # run automated tests
mvn clean         # remove generated files
mvn package       # create a JAR in target/
```

Lab 1 also uses `mvn exec:java`. Before using it, configure the Maven Exec Plugin and your main class in `pom.xml`; the [Gemini API Setup Guide](gemini_api_setup.ipynb) shows an example.

## Troubleshooting

- `mvn: command not found` or `mvn is not recognized`: reopen the terminal. If it persists, Maven is not on `PATH`; follow the manual setup steps on the [official installation page](https://maven.apache.org/install.html).
- `JAVA_HOME is not defined correctly`: install a JDK rather than a JRE, then configure `JAVA_HOME` to the JDK directory.
- Dependency download failures: confirm that Maven Central opens in your browser and ask campus IT for proxy or certificate guidance if you are on a restricted network. Do not commit the generated `target/` directory.
- Maven uses an unexpected Java version: run `mvn -version` to inspect the selected Java installation and correct `JAVA_HOME` if necessary.
